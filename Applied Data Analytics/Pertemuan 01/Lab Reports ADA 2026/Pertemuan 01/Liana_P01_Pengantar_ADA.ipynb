```python
# 1. Informasi tipe data setiap kolom
print("=== TIPE DATA ===")
print(df.dtypes)
print()

# 2. Cek nilai kosong
print("=== NILAI KOSONG (NaN) ===")
print(df.isnull().sum())
print()
```

```python
# 3. Statistik deskriptif (hanya kolom numerik)
print("=== STATISTIK DESKRIPTIF ===")
df.describe().round(2)
```

```python
# 4. BPRS dengan nasabah terbanyak
idx_max = df['total_nasabah'].idxmax()
bprs_terbesar = df.loc[idx_max, 'nama_bprs']
nasabah_terbanyak = df.loc[idx_max, 'total_nasabah']
print(f"🏆 BPRS terbesar: {bprs_terbesar} ({nasabah_terbanyak:,} nasabah)")

# 5. NPF tertinggi (BPRS paling berisiko)
idx_npf = df['npf_persen'].idxmax()
bprs_npf = df.loc[idx_npf, 'nama_bprs']
npf_val = df.loc[idx_npf, 'npf_persen']
print(f"⚠️  NPF tertinggi: {bprs_npf} ({npf_val}%)")

# 6. Rata-rata NPF industri (dari sampel ini)
avg_npf = df['npf_persen'].mean()
print(f"📊 Rata-rata NPF: {avg_npf:.2f}% (batas aman BI: 5%)")

# 7. Total pembiayaan seluruh BPRS
total = df['total_pembiayaan_juta'].sum()
print(f"💰 Total pembiayaan: Rp {total/1000:.1f} Miliar")
```

```python
df_aman = df[df['npf_persen'] < 3]

print(f"✅ Terdapat {df_aman.shape[0]} BPRS dengan NPF di bawah 3% (kategori aman).")
display(df_aman)
```

```python
df_sorted_nasabah = df.sort_values('total_nasabah', ascending=False)[['nama_bprs', 'total_nasabah']]
print("📊 BPRS dengan jumlah nasabah terbanyak:")
display(df_sorted_nasabah)
```

```python
df['pembiayaan_per_nasabah'] = df['total_pembiayaan_juta'] / df['total_nasabah']

idx_max_pembiayaan = df['pembiayaan_per_nasabah'].idxmax()
bprs_pembiayaan_tertinggi = df.loc[idx_max_pembiayaan, 'nama_bprs']
nilai_pembiayaan_tertinggi = df.loc[idx_max_pembiayaan, 'pembiayaan_per_nasabah']

print(f"⭐ BPRS dengan rasio pembiayaan per nasabah tertinggi adalah {bprs_pembiayaan_tertinggi} dengan nilai {nilai_pembiayaan_tertinggi:.2f} Juta/Nasabah.")
display(df[['nama_bprs', 'pembiayaan_per_nasabah']].sort_values(by='pembiayaan_per_nasabah', ascending=False).head())
```
